```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Shape of Dataset
print(df.shape)
```

```python
#Column Names
print(df.columns)
```

```python
#Dataset Information
print(df.info())
```

```python
#Check Missing Values
print(df.isnull().sum())
```

```python
#Drop Unnecessary Columns
Remove Unnamed: 32
df.drop('Unnamed: 32', axis=1, inplace=True)
```

```python
df.drop('id', axis=1, inplace=True)
```

```python
print(df.duplicated().sum())
```

```python
#Statistical Summary
print(df.describe())
```

```python
print(df['diagnosis'].value_counts())
```

```python
sns.countplot(x='diagnosis', data=df)

plt.title("Diagnosis Count")
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['diagnosis'] = le.fit_transform(df['diagnosis'])

# M = 1, B = 0
print(df.head())
```

```python
corr = df.corr()

print(corr)
```

```python
plt.figure(figsize=(20,15))

sns.heatmap(corr, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```python
df.hist(figsize=(20,15), bins=20)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20,15))

for i, col in enumerate(df.columns[1:], 1):
    plt.subplot(6, 5, i)
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
selected_features = [
    'radius_mean',
    'texture_mean',
    'perimeter_mean',
    'area_mean',
    'smoothness_mean',
    'diagnosis'
]

sns.pairplot(df[selected_features], hue='diagnosis')

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

outliers = ((numeric_df < (Q1 - 1.5 * IQR)) |
            (numeric_df > (Q3 + 1.5 * IQR)))

print(outliers.sum())
```

```python
sns.scatterplot(
    x='radius_mean',
    y='area_mean',
    hue='diagnosis',
    data=df
)

plt.title("Radius Mean vs Area Mean")
plt.show()
```

```python
diagnosis_counts = df['diagnosis'].value_counts()

print(diagnosis_counts)
```

```python
upper_triangle = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

high_corr = [
    column for column in upper_triangle.columns
    if any(upper_triangle[column] > 0.90)
]

print(high_corr)
```
